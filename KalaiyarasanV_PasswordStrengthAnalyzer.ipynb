# Password Strength Analyzer & Security Checker

**IBM SkillsBuild Cyber Security BootCamp 2026 – Project**

This project analyzes a password and classifies its strength based on length, uppercase/lowercase letters, digits, special characters, and common weak-password patterns.

```python
import re
import getpass

COMMON_PASSWORDS = {
    "password", "password123", "123456", "12345678", "123456789",
    "qwerty", "qwerty123", "admin", "admin123", "letmein",
    "welcome", "abc123", "iloveyou"
}

def analyze_password(password):
    score = 0
    checks = {
        "At least 8 characters": len(password) >= 8,
        "At least 12 characters": len(password) >= 12,
        "Contains uppercase letter": bool(re.search(r"[A-Z]", password)),
        "Contains lowercase letter": bool(re.search(r"[a-z]", password)),
        "Contains a number": bool(re.search(r"\d", password)),
        "Contains special character": bool(re.search(r"[^A-Za-z0-9]", password)),
        "Not a common password": password.lower() not in COMMON_PASSWORDS
    }

    if checks["At least 8 characters"]: score += 1
    if checks["At least 12 characters"]: score += 1
    if checks["Contains uppercase letter"]: score += 1
    if checks["Contains lowercase letter"]: score += 1
    if checks["Contains a number"]: score += 1
    if checks["Contains special character"]: score += 1
    if checks["Not a common password"]: score += 1

    if password.lower() in COMMON_PASSWORDS or score <= 2:
        strength = "Weak"
    elif score <= 4:
        strength = "Medium"
    elif score <= 6:
        strength = "Strong"
    else:
        strength = "Very Strong"

    suggestions = []
    if not checks["At least 8 characters"]:
        suggestions.append("Use at least 8 characters.")
    if not checks["At least 12 characters"]:
        suggestions.append("Prefer 12 or more characters.")
    if not checks["Contains uppercase letter"]:
        suggestions.append("Add uppercase letters.")
    if not checks["Contains lowercase letter"]:
        suggestions.append("Add lowercase letters.")
    if not checks["Contains a number"]:
        suggestions.append("Add numbers.")
    if not checks["Contains special character"]:
        suggestions.append("Add special characters such as !, @, #, or $.")
    if not checks["Not a common password"]:
        suggestions.append("Avoid common or easily guessed passwords.")

    return score, strength, checks, suggestions

password = getpass.getpass("Enter a password to analyze: ")
score, strength, checks, suggestions = analyze_password(password)

print("\nPASSWORD SECURITY REPORT")
print("-" * 30)
print(f"Score: {score}/7")
print(f"Strength: {strength}")
print("\nChecks:")
for item, passed in checks.items():
    print(f"{'✓' if passed else '✗'} {item}")

if suggestions:
    print("\nSuggestions:")
    for suggestion in suggestions:
        print("-", suggestion)
else:
    print("\nExcellent! The password meets all implemented checks.")
```

## Security Note

This tool is intended for educational password-security assessment. It does not store or transmit the password. For real accounts, use a unique password and a reputable password manager, and enable multi-factor authentication (MFA) where available.
